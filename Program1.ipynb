```python
x=int(input("Enter A Number1 : "))
y=int(input("Enter A Number2 : "))
print("**********************")
print("ARTHIMETIC OPERATORS")
print("Sum of Two Numbers : ",x+y)
print("Subtraction of Two Numbers : ",x-y)
print("Multiplication of Two Numbers : ",x*y)
print("Division of Two Numbers(DECIMAL DIVISION) : ",x/y)
print("Division of Two Numbers(INTEGER DIVISION) : ",x//y)
print("Remainder of Two Numbers: ",x%y)
print("**********************")
print("LOGICAL OPERATORS")
print(x>10 or y<5)
print(x>2 and y>3)
print(not x)
print("**********************")
print("COMPARISON OPERATORS")
if(x == y):
  print("Both Are Same Number")
elif(x>=y):
  print(f"X:{x} is Greater Than Y:{y}")
else:
   print(f"Y:{y} is Greater Than X:{x}")
print("**********************")
print("ASSIGNMENT OPERATORS")
print("Increment of X : ",x+1)
print("Decrement of Y : ",y-1)
print("**********************")
print("IDENTITY OPERATORS")
print(x is y)
print(x is not y)
print("**********************")



```
