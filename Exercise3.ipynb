```python
#List Functions
lst=[2,9,12,1,56,98,100]
print(lst)
print(len(lst))
print(sum(lst))
print(min(lst))
print(max(lst))
print(sorted(lst))
print(list(reversed(lst)))
```

```python
#List Methods
lst2=[1,98,23,33,12,56,90,87,66,21]
print(lst2)
lst2.append(100)
print(lst2)
lst2.insert(3,34)
print(lst2)
lst2.pop(2)
print(lst2)
lst.count(12)
print(lst2)
lst2.remove(21)
print(lst2)
lst2.extend(lst)
print(lst2)
lst2.sort()
print(lst2)
```

```python
#List Operations
lst3=[22,78,11,28,29,56,40]
lst4=[3,9,87,74,12,55,88]
print("Concatination : ",end="")
print(lst3+lst4)
print("Repetition: ",end="")
print(lst3*2)
print("Indexing :",end="")
print("Element at Position 3(Index 2): ",lst3[2])
print("Element at Position 3(Index 2) to Position 5(Index4): ",lst3[2:5])
print("Slicing : ",end="")
print("Reverse Of Elements Using Slicing: ",lst3[::-1])
print("Elements From The Last: ",lst3[-4:-1])
print("Membership Operators: ",end="")
print(3 in lst3)
print("Iterationns : ",end="")
for i in range(len(lst4)):
  print(lst4[i],end=" ")
```

```python
#Sum of the list without using functions
lst5=[22,78,11,28,29,56,40]
sum=0
for i in range(len(lst5)):
  sum+=lst5[i]
print(sum)
```

```python
#Sum of odd index in the list without using functions
lst5=[22,78,11,28,29,56,40]
sum=0
for i in range(len(lst5)):
  if(i%2!=0):
    sum+=lst5[i]

print(sum)
```

```python
#Sum of even index in the list without using functions
lst5=[22,78,11,28,29,56,40]
sum=0
for i in range(len(lst5)):
  if(lst5[i]%2==0):
    sum+=lst5[i]

print(sum)
```
